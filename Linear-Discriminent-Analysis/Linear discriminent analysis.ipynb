```python
#Standardization ensures that each feature contributes equally to the analysis. 
#Standardization can improve the stability and performance of the ld algorithm, especially in high dimesional columns where some features may dominate others
```

```python
sc = StandardScaler()
```

```python
x_train = sc.fit_transform(x_train)
```

```python
x_test = sc.transform(x_test)
```

```python
plt.scatter(x_train[:,0], x_train[:,1], c=y_train, cmap='viridis')
plt.title("LDA Projection")
plt.xlabel("LD1")
plt.ylabel("LD2")
plt.show()
```

```python
n = len(np.unique(y_train))
LDA = LinearDiscriminantAnalysis(n_components=n-1)
```

```python
# LDA
n = len(np.unique(y_train))
LDA = LinearDiscriminantAnalysis(n_components=2)

x_train = LDA.fit_transform(x_train, y_train)
x_test = LDA.transform(x_test)

# Model
log_model = LogisticRegression()
log_model.fit(x_train, y_train)

# Prediction
log_pred = log_model.predict(x_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, log_pred))
print(confusion_matrix(y_test, log_pred))
print(classification_report(y_test, log_pred))
```
