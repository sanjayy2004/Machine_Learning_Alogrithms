```python
#we need to find the relations between features (independent variables) and gain important information. to do that we will now find the covarience matric
```

```python
#COVARIENCE MATRIX
```

```python
X_T = X_scaled.T
pd.DataFrame(X_T)
```

```python
cov_mat = np.cov(X_T)
pd.DataFrame(cov_mat)
```

```python
#covarience matrix helps to find eigen values and vectors
```

```python
eig_values, eig_vectors = np.linalg.eig(cov_mat)
```

```python
pd.DataFrame(eig_values)
```

```python
pd.DataFrame(eig_vectors)
```

```python
#Explained varience : tells us how much information (vrience) each principle component captures
#more varience means it is explaining principle component , more important it is
```

```python
#HOW TO CALCULATE : to find out total information . we divide the eigen values by sum of all eigen values
```

```python
pd.DataFrame(eig_values)
```

```python
total_sum=sum(eig_values)
```

```python
total_sum
```

```python
sorted(eig_values, reverse=True)
```

```python
var_exp = []
for x in sorted(eig_values,reverse=True):
    exp=(x/total_sum)*100
    var_exp.append(exp)
var_exp
```

```python
#CUMILATIVE VARIENCE
#inorder to find the proper principle component and explained variance we caluculate cumulative varience.
#it helps us to decide how many components to keet. if we get 95% varience with first few components you can ignore the rest while still having most iomportant information
```

```python
cumulative_var_exp = np.cumsum(var_exp)
cumulative_var_exp
```

```python
cumulative_df = pd.DataFrame(cumulative_var_exp)
cumulative_df.head(30)
```

```python
plt.figure(figsize=(10,5))
plt.bar(range(len(var_exp)),var_exp,label='Individual_explained_varience',color='g')
plt.step(range(len(cumulative_var_exp)), cumulative_var_exp,label='Cumulative varience explained')
plt.xlabel('Explained Varience Ratio')
plt.ylabel('Explained varience ratio')
plt.legend()
plt.show()
```
